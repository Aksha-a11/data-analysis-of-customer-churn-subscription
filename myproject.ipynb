```python
import matplotlib.pyplot as plt
import seaborn as sns 
import pandas as pd 
import numpy as np 
df=pd.read_csv('C:/Users/Hp/Downloads/mydata/customer_subscription_churn_usage_patterns.csv')


```

```python
df["signup_date"] = pd.to_datetime(df["signup_date"], dayfirst=True)
```

```python
df["signup_year"] = df["signup_date"].dt.year

sns.countplot(data=df, x="signup_year")
plt.show()
```

```python

df["signup_month"] = df["signup_date"].dt.month_name()

plt.figure(figsize=(12,5))
sns.countplot(data=df, x="signup_month",
              order=['January','February','March','April','May','June',
                     'July','August','September','October','November','December'])
plt.xticks(rotation=45)
plt.show()
```

```python
churn_counts = df['churn'].value_counts()

plt.figure(figsize=(6,6))

plt.pie(churn_counts.values,
        labels=churn_counts.index,
        autopct='%1.1f%%',
        startangle=90)

plt.title("Customer Churn Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,6))

plt.scatter(df['monthly_fee'],
            df['avg_weekly_usage_hours'])

plt.title("Monthly Fee vs Weekly Usage")
plt.xlabel("Monthly Fee")
plt.ylabel("Weekly Usage Hours")

plt.show()
```

```python
avg_fee = df.groupby('plan_type')['monthly_fee'].mean()

plt.figure(figsize=(7,5))

plt.plot(avg_fee.index,
         avg_fee.values,
         marker='o')

plt.title("Average Monthly Fee by Plan")
plt.xlabel("Plan Type")
plt.ylabel("Average Fee")

plt.show()
```

```python
support = df.groupby('plan_type')['support_tickets'].mean()

plt.figure(figsize=(7,5))

plt.bar(support.index,
        support.values)

plt.title("Average Support Tickets")
plt.xlabel("Plan Type")
plt.ylabel("Average Tickets")

plt.show()
```

```python
payment = df['payment_failures'].value_counts().sort_index()

plt.figure(figsize=(7,5))

plt.bar(payment.index,
        payment.values)

plt.title("Payment Failures")
plt.xlabel("Number of Failures")
plt.ylabel("Customers")

plt.show()
```

```python
tickets = df['support_tickets'].value_counts().sort_index()

plt.figure(figsize=(8,5))

plt.bar(tickets.index,
        tickets.values)

plt.title("Support Tickets")
plt.xlabel("Tickets")
plt.ylabel("Customers")

plt.show()
```

```python
plt.figure(figsize=(6,5))

plt.boxplot(df['tenure_months'])

plt.title("Tenure Boxplot")

plt.show()
```

```python
# Dataset Description

## Overview

The **Customer Subscription Churn Usage Patterns** dataset contains information about customers who have subscribed to a service. It includes customer demographics related to their subscription, usage behavior, payment history, customer support interactions, and churn status. The primary objective of this dataset is to analyze the factors that influence customer churn and identify patterns that can help businesses improve customer retention.

The dataset consists of **2,800 customer records** and **10 features**, where each row represents one customer and each column represents a specific attribute of that customer.

## Features Description

| Feature                    | Description                                                                                                                                    |
| -------------------------- | ---------------------------------------------------------------------------------------------------------------------------------------------- |
| **user_id**                | A unique identification number assigned to each customer. It is used only to identify records and does not influence churn prediction.         |
| **signup_date**            | The date on which the customer subscribed to the service. This helps analyze customer acquisition trends over time.                            |
| **plan_type**              | The subscription plan selected by the customer (such as Basic, Standard, or Premium). Different plans may have different pricing and features. |
| **monthly_fee**            | The monthly subscription amount paid by the customer for the selected plan.                                                                    |
| **avg_weekly_usage_hours** | The average number of hours the customer uses the service each week. It represents customer engagement.                                        |
| **support_tickets**        | The number of customer support requests raised by the customer. A higher number may indicate dissatisfaction or service-related issues.        |
| **payment_failures**       | The number of times the customer's payment failed. Frequent payment failures may increase the likelihood of churn.                             |
| **tenure_months**          | The total number of months the customer has remained subscribed to the service. It indicates customer loyalty.                                 |
| **last_login_days_ago**    | The number of days since the customer's last login. Larger values may indicate inactive customers.                                             |
| **churn**                  | The target variable that indicates whether the customer has left the service (**Yes**) or is still an active subscriber (**No**).              |

## Purpose of the Dataset

The main purpose of this dataset is to understand customer behavior and identify the key factors that contribute to customer churn. By analyzing usage patterns, subscription plans, payment history, customer support interactions, and customer activity, organizations can identify customers who are at risk of leaving the service.

The insights obtained from this dataset can help businesses:

* Improve customer retention strategies.
* Understand customer engagement patterns.
* Reduce customer churn.
* Enhance customer satisfaction.
* Support data-driven business decisions.
* Build machine learning models for churn prediction.

## Target Variable

The target variable in this dataset is **`churn`**, which is a binary categorical variable:

* **Yes** – The customer has discontinued the subscription.
* **No** – The customer continues to use the subscription service.

This makes the dataset suitable for **binary classification** problems in machine learning.

## Applications

This dataset can be used for:

* Exploratory Data Analysis (EDA)
* Customer Churn Prediction
* Customer Segmentation
* Business Intelligence Dashboards
* Predictive Analytics
* Machine Learning Classification Models
* Power BI Dashboard Development
```
